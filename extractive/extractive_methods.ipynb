```python
def clean_khmer_text(text):
    if pd.isna(text):
        return ""
    text = str(text)
    text = text.replace("\u200b", "").replace("\ufeff", "")
    return text.strip()

def split_khmer_sentences(text):
    text = clean_khmer_text(text)
    sentences = text.split("។")
    return [s.strip() for s in sentences if len(s.strip()) > 5]
```

Study and compare three extractive summarization methods:

- TF-IDF Ranking
- TextRank
- Clustering-Based Extraction

All methods:

- Use the same cleaned text
- Work at sentence level
- Output an extractive summary
- Differ in how sentences are selected

1. TF-IDF Ranking 
    - Rank sentences by keyword importance

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import numpy as np

# TF-IDF Vectorization
vectorizer = TfidfVectorizer(token_pattern=r"(?u)\b\w+\b")
tfidf_matrix = vectorizer.fit_transform(sentences)

#TF-IDF Ranking
sentence_scores = np.array(tfidf_matrix.sum(axis=1)).ravel()

top_n = min(2, len(sentences))
top_indices = sentence_scores.argsort()[-top_n:][::-1]

tfidf_summary = "។ ".join([sentences[i] for i in top_indices]) + "។"
```

```python
print("TF-IDF MATRIX SHAPE:")
print(tfidf_matrix.shape)

print("\nTF-IDF FEATURE EXAMPLE:")
print(vectorizer.get_feature_names_out()[:100])
```

```python
print("SENTENCE SCORES:")
for i, score in enumerate(sentence_scores):
    print(f"Sentence {i+1}: Score = {score:.4f}")
```

- If ratio = 30% and you have 4 sentences:
- Select top 1–2 sentences with highest important score
- Result: copy senttences, keep original wording


```python
# Sentence Ranking & Selection
ratio = 0.3  
ranked = sorted(
    ((score, sent, idx) for idx, (score, sent) in enumerate(zip(sentence_scores, sentences))),
    reverse=True
)

k = max(1, int(len(sentences) * ratio))
selected = ranked[:k]
selected = sorted(selected, key=lambda x: x[2])

print("SELECTED SENTENCES:")
for score, sent, idx in selected:
    print(f"[Score: {score:.4f}] {sent}")
```

```python
# Apply TF-IDF to Whole DataFrame 
def extract_summary_tfidf(article, ratio=0.3):
    sentences = split_khmer_sentences(article)
    if not sentences:
        return ""
    tfidf = TfidfVectorizer(token_pattern=r"(?u)\b\w+\b").fit_transform(sentences)
    scores = np.array(tfidf.sum(axis=1)).ravel()
    ranked = sorted(
        ((score, sent, idx) for idx, (score, sent) in enumerate(zip(scores, sentences))),
        reverse=True
    )
    k = max(1, int(len(sentences) * ratio))
    selected = sorted(ranked[:k], key=lambda x: x[2])
    return "។ ".join([s[1] for s in selected]) + "។"
```

2. TextRank
    - Rank sentences using a graph of sentence similarity
(similar to PageRank)

```python
from sklearn.metrics.pairwise import cosine_similarity

def extract_summary_textrank(article, ratio=0.3):
    sentences = split_khmer_sentences(article)
    if not sentences:
        return ""

    tfidf = TfidfVectorizer(token_pattern=r"(?u)\b\w+\b").fit_transform(sentences)

    similarity_matrix = cosine_similarity(tfidf)
    scores = similarity_matrix.sum(axis=1)

    ranked = sorted(
        ((score, sent, idx) for idx, (score, sent) in enumerate(zip(scores, sentences))),
        reverse=True
    )

    k = max(1, int(len(sentences) * ratio))
    selected = sorted(ranked[:k], key=lambda x: x[2])

    return "။ ".join([s[1] for s in selected]) + "។"
```

3. Clustering-Based Extraction 
    - Group similar sentences → pick representatives

```python
from sklearn.cluster import KMeans
from sklearn.metrics.pairwise import cosine_similarity

def extract_summary_cluster(article, ratio=0.3):
    sentences = split_khmer_sentences(article)
    if not sentences:
        return ""

    tfidf = TfidfVectorizer(token_pattern=r"(?u)\b\w+\b").fit_transform(sentences)

    k = max(1, int(len(sentences) * ratio))
    k = min(k, len(sentences))  # safety

    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(tfidf)
    labels = kmeans.labels_

    summary_sentences = []

    for cluster_id in range(k):
        idx = np.where(labels == cluster_id)[0]
        cluster_vecs = tfidf[idx]
        centroid = kmeans.cluster_centers_[cluster_id]

        sims = cosine_similarity(cluster_vecs, centroid.reshape(1, -1))
        best_idx = idx[np.argmax(sims)]

        summary_sentences.append((best_idx, sentences[best_idx]))

    # keep original order
    summary_sentences = sorted(summary_sentences, key=lambda x: x[0])

    return "។ ".join([s[1] for s in summary_sentences]) + "។"
```

- Compare results

```python
# Select the row
row = df.iloc[5]
article = row["Article"]

# Print original article
print("1️ ORIGINAL ARTICLE:\n")
print(article)

# Print the labeled summary from your DataFrame
print("\n2️ LABELED SUMMARY (from DataFrame):\n")
print(row["Summary"])

# Generate summaries with the 3 methods
textrank_summary = extract_summary_textrank(article)
tfidf_summary = extract_summary_tfidf(article)
cluster_summary = extract_summary_cluster(article)

# Print summaries from each method
print("\n3️ TEXT RANK SUMMARY:\n", textrank_summary)
print("\n4️ TF-IDF SUMMARY:\n", tfidf_summary)
print("\n5️ CLUSTERING SUMMARY:\n", cluster_summary)
```

### Cosine similarity Evaluator

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

summaries = [textrank_summary, tfidf_summary, cluster_summary]
vectorizer = TfidfVectorizer().fit([article] + summaries)

article_vec = vectorizer.transform([article])
summary_vecs = vectorizer.transform(summaries)

for i, summary in enumerate(["TextRank", "TF-IDF", "Clustering"]):
    sim = cosine_similarity(article_vec, summary_vecs[i])[0][0]
    print(f"{summary} similarity to original article: {sim:.2f}")
```
